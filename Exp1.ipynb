1(a) Python Tokens and Keywords

```python

age = 20

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")

name = "Asha"
number = 25
price = 3.14
message = "Hello Python"
result = number + 10

print("Name:", name)
print("Number:", number)
print("Price:", price)
print("Result:", result)
```

1(b) Variable Declaration and Dynamic Typing

```python

value = 100
print("Value:", value)
print("Type:", type(value))

value = 25.5
print("\nValue:", value)
print("Type:", type(value))

value = "Python"
print("\nValue:", value)
print("Type:", type(value))
```

1(c) Basic Data Types

```python

integer_value = 25
float_value = 12.5
string_value = "Python"
boolean_value = True

print("Integer:", integer_value)
print("Type:", type(integer_value))

print("\nFloat:", float_value)
print("Type:", type(float_value))

print("\nString:", string_value)
print("Type:", type(string_value))

print("\nBoolean:", boolean_value)
print("Type:", type(boolean_value))
```

2. Arithmetic, Relational, Logical and Assignment Operations

```python
a = 20
b = 6

print("ARITHMETIC OPERATIONS")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

print("\nRELATIONAL OPERATIONS")
print("a > b:", a > b)
print("a < b:", a < b)
print("a == b:", a == b)
print("a != b:", a != b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)

x = True
y = False

print("\nLOGICAL OPERATIONS")
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)

number = 10

print("\nASSIGNMENT OPERATIONS")
number += 5
print("After += 5:", number)

number -= 3
print("After -= 3:", number)

number *= 2
print("After *= 2:", number)

number //= 2
print("After //= 2:", number)
```

3. Operator Precedence
Python follows a specific order while evaluating expressions.

```python

a = 10
b = 5
c = 2

result = a + b * c

print("Expression: a + b * c")
print("Result:", result)

result2 = (a + b) * c

print("\nExpression: (a + b) * c")
print("Result:", result2)

result3 = a + b * c ** 2

print("\nExpression: a + b * c ** 2")
print("Result:", result3)
```

4. Type Casting Functions

```python

number_string = "25"
decimal_string = "12.5"

integer_value = int(number_string)
float_value = float(decimal_string)
string_value = str(100)
boolean_value = bool(1)

print("Original:", number_string, type(number_string))
print("After int():", integer_value, type(integer_value))

print("\nOriginal:", decimal_string, type(decimal_string))
print("After float():", float_value, type(float_value))

print("\nAfter str():", string_value, type(string_value))
print("After bool():", boolean_value, type(boolean_value))
```

5. Using input() for Basic Computations

```python

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\n--- Results ---")
print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)

if num2 != 0:
    print("Division:", num1 / num2)
else:
    print("Division is not possible by zero.")
```
