#EXPERIMENT 3: Functions and Recursion

1. User-Defined Function

```python
# User-defined function

def calculate_area(length, breadth):
    area = length * breadth
    return area


length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))

area = calculate_area(length, breadth)

print("Area of rectangle:", area)
```

2(a). Default Arguments

```python
# Demonstration of default arguments

def greet(name, message="Welcome to Python"):
    print(name + ", " + message)


greet("Asha")
greet("Rahul", "Good morning")
```

2(b). Keyword Arguments

```python
# Demonstration of keyword arguments

def student_details(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)


student_details(
    course="BTech CSE",
    name="Asha",
    age=20
)
```

2(c). Variable-Length Arguments

```python
# Variable-length positional arguments

def calculate_sum(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total


print("Sum:", calculate_sum(10, 20))
print("Sum:", calculate_sum(10, 20, 30, 40))
```

```python
# Variable-length keyword arguments

def display_details(**details):
    for key, value in details.items():
        print(key, ":", value)


display_details(
    name="Asha",
    age=20,
    course="BTech CSE"
)
```

3(a). Lambda with map()

```python
# Square every number using lambda and map()

numbers = [1, 2, 3, 4, 5]

squares = list(map(lambda x: x ** 2, numbers))

print("Numbers:", numbers)
print("Squares:", squares)
```

3(b). Lambda with filter()

```python
# Filter even numbers using lambda and filter()

numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print("Numbers:", numbers)
print("Even numbers:", even_numbers)
```

4(a). Recursive Factorial

```python
# Recursive program to calculate factorial

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


number = int(input("Enter a number: "))

if number >= 0:
    print("Factorial:", factorial(number))
else:
    print("Factorial is not defined for negative numbers.")
```

4(b). Recursive Fibonacci

```python
# Recursive program to generate Fibonacci numbers

def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)


terms = int(input("Enter number of terms: "))

print("Fibonacci Series:")

for i in range(terms):
    print(fibonacci(i), end=" ")
```

5. Small Application Using Modular Functions

```python
# Student Grade Calculator using modular functions

def calculate_average(marks):
    return sum(marks) / len(marks)


def calculate_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 40:
        return "D"
    else:
        return "F"


def display_result(name, marks):
    average = calculate_average(marks)
    grade = calculate_grade(average)

    print("\n===== RESULT =====")
    print("Name:", name)
    print("Marks:", marks)
    print("Average:", round(average, 2))
    print("Grade:", grade)


name = input("Enter student name: ")

marks = []

for i in range(3):
    mark = float(input(f"Enter marks for subject {i + 1}: "))
    marks.append(mark)

display_result(name, marks)
```
