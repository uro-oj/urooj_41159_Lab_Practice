#EXPERIMENT 5: File Handling and Exception Handling

1. File Access Modes — r, w, a

```python
# Demonstration of file access modes

# Write mode
file = open("sample.txt", "w")
file.write("Python Programming\n")
file.write("File Handling\n")
file.close()

# Append mode
file = open("sample.txt", "a")
file.write("Exception Handling\n")
file.close()

# Read mode
file = open("sample.txt", "r")
content = file.read()
file.close()

print("File contents:")
print(content)
```

2(a). Write Data to a File

```python
# Writing data to a file

file = open("students.txt", "w")

file.write("Asha - 85\n")
file.write("Rahul - 91\n")
file.write("Priya - 78\n")

file.close()

print("Data written successfully.")
```

2(b). Read Data from a File

```python
# Reading data from a file

file = open("students.txt", "r")

data = file.read()

file.close()

print("Student Records:")
print(data)
```

2(c). Using the with Statement

```python
# Using with statement

with open("students.txt", "r") as file:
    data = file.read()

print("File contents:")
print(data)
```

3. Working with Directories Using os

```python
import os

directory = "LabFiles"

# Create directory
if not os.path.exists(directory):
    os.mkdir(directory)
    print("Directory created.")
else:
    print("Directory already exists.")

# Display current directory
print("Current working directory:")
print(os.getcwd())

# Display directory contents
print("\nDirectory contents:")
print(os.listdir())
```

4(a). try-except-else-finally

```python
# Demonstration of try-except-else-finally

try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Program execution completed.")
```

4(b). Handling Multiple Exceptions

```python
# Handling multiple exceptions

try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2

    print("Result:", result)

except ValueError:
    print("Error: Please enter integers only.")

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

(c). raise and assert

```python
# Demonstration of raise

age = int(input("Enter your age: "))

try:
    if age < 18:
        raise ValueError("Age must be 18 or above.")

    print("Eligible.")

except ValueError as error:
    print("Error:", error)
```

```python
# Demonstration of assert

marks = int(input("Enter marks: "))

try:
    assert 0 <= marks <= 100, "Marks must be between 0 and 100."
    print("Valid marks:", marks)

except AssertionError as error:
    print("Error:", error)
```

5. Custom Exception

```python
# Custom exception demonstration

class InsufficientMarksError(Exception):
    pass


try:
    marks = int(input("Enter marks: "))

    if marks < 40:
        raise InsufficientMarksError(
            "Student has not achieved the passing marks."
        )

    print("Student has passed.")

except InsufficientMarksError as error:
    print("Custom Exception:", error)
```
