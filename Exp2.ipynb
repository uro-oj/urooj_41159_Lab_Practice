1(a) if, if-else and Nested if

```python


marks = int(input("Enter your marks: "))

if marks >= 40:
    print("You have passed.")

if marks >= 50:
    print("Grade: Pass")
else:
    print("Grade: Needs improvement")

if marks >= 40:
    if marks >= 75:
        print("Performance: Distinction")
    else:
        print("Performance: Good")
else:
    print("Performance: Not satisfactory")
```

1(b) for and while Loops

```python

print("FOR LOOP")

for i in range(1, 6):
    print(i)


print("\nWHILE LOOP")

number = 1

while number <= 5:
    print(number)
    number += 1
```

2(a) Lists

```python
# Demonstration of list operations

fruits = ["Apple", "Banana", "Mango"]

print("Original list:", fruits)

# append
fruits.append("Orange")
print("After append:", fruits)

# insert
fruits.insert(1, "Grapes")
print("After insert:", fruits)

# remove
fruits.remove("Banana")
print("After remove:", fruits)

# slicing
print("First three elements:", fruits[:3])
print("Last two elements:", fruits[-2:])
```

2(b) Tuples

```python
# Demonstration of tuple indexing and immutability

student = ("Asha", 20, "BTech", 85.5)

print("Tuple:", student)

# Indexing
print("Name:", student[0])
print("Age:", student[1])
print("Course:", student[2])

# Tuples are immutable
try:
    student[1] = 21
except TypeError:
    print("\nTuple elements cannot be changed.")
```

2(c) Sets

```python
# Demonstration of set operations

set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)

print("\nUnion:", set_a.union(set_b))
print("Intersection:", set_a.intersection(set_b))
print("Difference A - B:", set_a.difference(set_b))
print("Difference B - A:", set_b.difference(set_a))
```

2(d) Dictionaries

```python
# Demonstration of dictionary operations

student = {
    "name": "Asha",
    "age": 20,
    "course": "BTech"
}

print("Original dictionary:", student)

# Add
student["marks"] = 85
print("\nAfter adding marks:", student)

# Update
student["age"] = 21
print("After updating age:", student)

# Delete
del student["course"]
print("After deleting course:", student)
```

3. Student Record System

```python
# Simple Student Record System

students = {}

number_of_students = int(input("Enter number of students: "))

for i in range(number_of_students):
    print(f"\nEnter details for Student {i + 1}")

    roll_no = input("Roll number: ")
    name = input("Name: ")
    marks = float(input("Marks: "))

    students[roll_no] = {
        "name": name,
        "marks": marks
    }

print("\n===== STUDENT RECORDS =====")

for roll_no, details in students.items():
    print("Roll No:", roll_no)
    print("Name:", details["name"])
    print("Marks:", details["marks"])

    if details["marks"] >= 40:
        print("Result: Pass")
    else:
        print("Result: Fail")

    print("-" * 30)
```
