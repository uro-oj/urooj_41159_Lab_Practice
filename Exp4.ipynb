4(a). Validate Email Address Using Regex

```python
import re

email = input("Enter email address: ")

pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'

if re.match(pattern, email):
    print("Valid email address")
else:
    print("Invalid email address")
```

4(b). Extract Mobile Numbers from Text

```python
import re

text = """
Contact Asha at 9876543210.
Rahul's number is 9123456789.
"""

pattern = r'\b[6-9]\d{9}\b'

mobile_numbers = re.findall(pattern, text)

print("Mobile numbers found:")

for number in mobile_numbers:
    print(number)
```

4(c). Regular Expression Metacharacters

```python
import re

text = "Python 123 is powerful. Python 456 is easy."

# .  -> Any character
# ^  -> Beginning of string
# $  -> End of string
# \d -> Digit
# \w -> Word character
# +  -> One or more occurrences
# *  -> Zero or more occurrences

print("Digits:", re.findall(r'\d+', text))
print("Words:", re.findall(r'\w+', text))
print("Python occurrences:", re.findall(r'Python', text))
```
